```python
# Extract products currently in stock, calculate each product's inventory value, and calculate the total inventory value.

from functools import reduce

products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

filtered = filter(lambda x: x["stock"] > 0, products)

values = list(map(lambda x: x["price"] * x["stock"], filtered))

result = reduce(lambda x, y: x + y, values)

print(result)
```

```python
# Extract Electronics products, apply a 20% discount, and calculate the total discounted value.

from functools import reduce

products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]

filtered = filter(lambda x: x["category"] == "Electronics", products)

discounted = list(map(lambda x: x["price"] * 0.80, filtered))

result = reduce(lambda x, y: x + y, discounted)

print(result)
```

```python
# Extract delivered orders and calculate the total delivered revenue.

from functools import reduce

orders = [
    {"order_id": 1001, "amount": 1200, "status": "Delivered"},
    {"order_id": 1002, "amount": 3500, "status": "Pending"},
    {"order_id": 1003, "amount": 8500, "status": "Delivered"},
    {"order_id": 1004, "amount": 4200, "status": "Cancelled"},
    {"order_id": 1005, "amount": 7600, "status": "Delivered"},
    {"order_id": 1006, "amount": 5100, "status": "Pending"}
]

filtered = filter(lambda x: x["status"] == "Delivered", orders)

result = reduce(lambda total, order: total + order["amount"], filtered, 0)

print(result)
```

```python
# Extract transactions greater than 5000, add 18% GST, and calculate the grand total after GST.

from functools import reduce

transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

filtered = filter(lambda x: x["amount"] > 5000, transactions)

gst_amount = list(map(lambda x: x["amount"] * 1.18, filtered))

result = reduce(lambda x, y: x + y, gst_amount)

print(result)
```

```python
# Extract active salespeople from the South region, calculate a 5% incentive, create a list with name, sales, and incentive, and calculate the total incentive payable.

from functools import reduce

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

filtered = filter(lambda x: x["region"] == "South" and x["active"], sales_records)

details = list(map(lambda x: {
    "salesperson": x["salesperson"],
    "sales": x["sales"],
    "incentive": x["sales"] * 0.05
}, filtered))

print(details)

total_incentive = reduce(lambda total, x: total + x["incentive"], details, 0)

print(total_incentive)
```
